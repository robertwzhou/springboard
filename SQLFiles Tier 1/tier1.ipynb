```python
import sqlite3
from sqlite3 import Error

 
def create_connection(db_file):
    """ create a database connection to the SQLite database
        specified by the db_file
    :param db_file: database file
    :return: Connection object or None
    """
    conn = None
    try:
        conn = sqlite3.connect(db_file)
        print(sqlite3.version)
    except Error as e:
        print(e)
 
    return conn

 
def select_all_tasks(conn):
    """
    Query all rows in the tasks table
    :param conn: the Connection object
    :return:
    """
    cur = conn.cursor()
    
    query1 = """
        SELECT *
        FROM FACILITIES
        """
    cur.execute(query1)
 
    rows = cur.fetchall()
 
    for row in rows:
        print(row)


def main():
    database = "sqlite_db_pythonsqlite.db"
 
    # create a database connection
    conn = create_connection(database)
    with conn: 
        print("2. Query all tasks")
        select_all_tasks(conn)
 
 
if __name__ == '__main__':
    main()
```

```python
def q(query):
    cur = conn.cursor()
    
    cur.execute(query)
 
    rows = cur.fetchall()
 
    for row in rows:
        print(row)
```

```python
#Q1
facilities_with_member_fee = """
SELECT name
FROM facilities
WHERE membercost > 0
"""
q(facilities_with_member_fee)
```

```python
#Q2
num_facilities_without_member_fee = """
SELECT COUNT(*)
FROM facilities
WHERE membercost = 0
"""
q(num_facilities_without_member_fee)
```

```python
#Q3
facilities_where_membercost_lt_20percent_monthlymaintenance = """
SELECT facid, name, membercost, monthlymaintenance
FROM facilities
WHERE 0 < membercost AND membercost < 0.2 * monthlymaintenance
"""
q(facilities_where_membercost_lt_20percent_monthlymaintenance)
```

```python
#Q4
facilities_1_and_5 = """
SELECT *
FROM facilities
WHERE facid IN (1, 5)
"""
q(facilities_1_and_5)
```

```python
#Q5
facilities_cheap_expensive = """
SELECT
    name,
    monthlymaintenance,
    CASE WHEN monthlymaintenance > 100 THEN 'expensive'
        ELSE 'cheap' END AS label
FROM facilities
"""
q(facilities_cheap_expensive)
```

```python
#Q6
name_of_last_signup = """
SELECT firstname, surname AS lastname
FROM members
WHERE joindate = (
    SELECT MAX(joindate)
    FROM members
)
"""
q(name_of_last_signup)
```

```python
#Q7
tennis_court_users = """
SELECT DISTINCT f.name AS tenniscourt_name, m.firstname || ' ' || m.surname AS membername
FROM members AS m
INNER JOIN bookings AS b USING(memid)
INNER JOIN facilities AS f USING(facid)
WHERE f.name LIKE 'Tennis Court%'
ORDER BY membername
"""
q(tennis_court_users)
```

```python
q8 = """
SELECT
    f.name,
    m.firstname || ' ' || m.surname AS membername,
    CASE WHEN m.memid = 0 THEN f.guestcost
        ELSE f.membercost END AS cost
FROM members AS m
INNER JOIN bookings AS b USING(memid)
INNER JOIN facilities AS f USING(facid)
WHERE b.starttime LIKE '2012-09-14%'
    AND ((m.memid = 0 AND f.guestcost > 30)
        OR (m.memid <> 0 AND f.membercost > 30))
ORDER BY cost DESC
"""
q(q8)
```

```python
q9 = """
SELECT name, membername, cost
FROM
    (SELECT f.name,
    m.firstname || ' ' || m.surname AS membername,
    CASE WHEN m.memid = 0 THEN f.guestcost
        ELSE f.membercost END AS cost
    FROM members AS m
    INNER JOIN bookings AS b USING(memid)
    INNER JOIN facilities AS f USING(facid)
    WHERE b.starttime LIKE '2012-09-14%'
        AND ((m.memid = 0 AND f.guestcost > 30)
        OR (m.memid <> 0 AND f.membercost > 30)))
ORDER BY cost DESC
"""
q(q9)
```

```python
#Q10
facilities_with_total_revenue_lt1000 = """
SELECT name, SUM(cost) AS revenue
FROM (
    SELECT f.name,
        CASE WHEN b.memid = 0 THEN f.guestcost
            ELSE f.membercost END AS cost
    FROM bookings AS b
    INNER JOIN facilities AS f USING(facid)
) AS booking_costs
GROUP BY name
HAVING SUM(cost) < 1000
ORDER BY revenue
"""
q(facilities_with_total_revenue_lt1000)
```

```python
#Q11
members_recommenders = """
SELECT DISTINCT m.surname || ',' || m.firstname AS member,
    r.surname || ',' || r.firstname AS recommender
FROM members as m
LEFT JOIN members AS r ON m.recommendedby = CAST(r.memid AS VARCHAR(2))
ORDER BY member, recommender
"""
q(members_recommenders)
```

```python
facilities_usage_by_member = """
WITH usages AS (SELECT facid, memid, COUNT(*) AS usage
FROM bookings AS b
WHERE memid <> 0
GROUP BY facid, memid)
SELECT f.name, m.firstname || ' ' || m.surname AS membername, usage
FROM usages
INNER JOIN facilities AS f USING(facid)
INNER JOIN members AS m USING(memid)
"""
q(facilities_usage_by_member)
```

```python
facilities_usage_by_month = """
WITH usages AS (SELECT facid, strftime('%m', starttime) AS month, COUNT(*) AS usage
FROM bookings AS b
WHERE memid <> 0
GROUP BY facid, memid)
SELECT f.name, month, usage
FROM usages
INNER JOIN facilities AS f USING(facid)
"""
q(facilities_usage_by_month)
```
